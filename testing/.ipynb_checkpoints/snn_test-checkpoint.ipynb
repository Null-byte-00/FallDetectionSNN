# Simple LIF Network

```python
import torch
import torch.nn as nn



class LIFLayer(nn.Module):
    def __init__(self, input_size, num_neurons,
                 beta=0.9, threshold=1.0):

        super().__init__()

        self.synapses = nn.Linear(
            input_size,
            num_neurons,
            bias=False
        )

        self.beta = beta
        self.threshold = threshold

    def forward(self, input_spikes, membrane):

        current = self.synapses(input_spikes)

        membrane = self.beta * membrane + current

        spikes = (membrane >= self.threshold).float()

        membrane = membrane * (1 - spikes)

        return spikes, membrane

STEPS = 10

layer = LIFLayer(
    input_size=3,
    num_neurons=3
)

membrane = torch.zeros(3)

for i in range(100):
    inp = torch.rand(3);
    spikes, membrane = layer(inp, membrane);
    for j in range(3):
        if spikes[j] == 0:
            print("_")
        else:
            print(1)
    print("-*-*-*")
```
